```python
from pathlib import Path
import urllib.request, numpy as np, pandas as pd, matplotlib.pyplot as plt
import trimesh

STL_URL = "https://raw.githubusercontent.com/BosenkoTM/robotics-course/2026/06-simulation-and-sim2real/seminar/so101/Wrist_Cam_Mount_32x32_UVC_Module_SO101.stl"
stl_path = Path("so101_wrist_cam_mount.stl")
if not stl_path.exists():
    urllib.request.urlretrieve(STL_URL, stl_path)
mesh = trimesh.load_mesh(stl_path)
mesh
```

```python
def variant_params(n):
    mats=["PLA","PETG","ABS","TPU","PLA"]
    densities={"PLA":1.24,"PETG":1.27,"ABS":1.04,"TPU":1.20} # g/cm3, учебная оценка
    layers=[0.10,0.12,0.16,0.20,0.24]
    return {
        "material": mats[(n-1)%5],
        "density": densities[mats[(n-1)%5]],
        "layer_height_mm": layers[(n-1)//5],
        "infill_pct": [10,15,20,25,30][(n-1)%5],
        "scale": 0.90+0.01*(n-1),
        "orientation": ["X","Y","Z","X","Y"][(n-1)%5],
    }
cfg=variant_params(VARIANT)
cfg
```
